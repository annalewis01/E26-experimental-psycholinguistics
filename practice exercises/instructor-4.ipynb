```python
branches_of_linguistics = ["phonology", "semiotics", "morphology", "phonetics", "pragmatics", "syntax", "sociolinguistics", "psycholinguistics", "computational linguistics", "applied linguistics"]
```

```python
for branch in branches_of_linguistics:
    # print(branch.upper() + len(branch))
    # this will not work because you can't add a string and an integer. print gives a string (the word) whilst len gives an integer (length)
    # solution -> make the length into a string
    # print(branch.upper() + str(len(branch)))
    # but this is hard to read so we add a space
    print(branch.upper() + " " + str(len(branch)))
```

```python
for branch in branches_of_linguistics:
    print(branch.upper(), len(branch))

# this works too, but here the length is still an integer, not a string. this is fine i think
```
