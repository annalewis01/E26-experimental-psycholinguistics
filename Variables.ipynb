From https://www.w3schools.com:

- **Strings** are for storing text
- **Integers** are for storing whole numbers, positive or negative (such as 123 or -456), without decimals.
- **Floats** are for storing positive and negative numbers with a decimal point, like 35.3, or -2.34
- **Lists** are collections which are ordered and changeable. Allows duplicate members.
- **Tuples** are collections which are ordered and unchangeable. Allows duplicate members.
- **Sets** are collections which are unordered, "unchangeable", and unindexed. No duplicate members.
- **Dictionaries** are collections which are ordered and changeable. No duplicate members.

## Sets

```python
newList = [4, 4, 7, 19, 21, "i am a string", 7.23]
mySet = set(newList)
mySet
```

no duplicate members in sets. also unordered and unchangeable. somewhat random order but this is not that important

sets are useful for looking for types vs tokens - e.g. a text with lots of duplicate words where you don't care about how many times a word appears but are looking at how many different words there are

```python
mySet[3]
```

no fixed order in a set, therefore you can't isolate an item

```python
anotherSet = {9, 12, 15}
anotherSet
```

```python
type(anotherSet)
```

```python
oneMoreSet = set(29, 40, 6)
```

you can't make a set using the set function with more than one argument for some reason. you can make a set out of a list (i.e. the data for the set is gathered first in a list) or you can make a set directly from the data using curly brackets.
